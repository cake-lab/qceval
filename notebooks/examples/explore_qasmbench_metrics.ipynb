```python
from qceval import CircuitLoader, CircuitMetricsEvaluator

loader = CircuitLoader()
evaluator = CircuitMetricsEvaluator()
```

```python
circuits = list(loader.load("disqco.mfpqc.qaoa.n16"))[0:2]

circuits_decomposed = [(name, circuit.decompose(reps=10)) for name, circuit in circuits]

circuit_a_id, circuit_a = next(loader.load("qasmbench.small.adder_n4"))
circuit_b_id, circuit_b = next(loader.load("qasmbench.small.bell_n4"))
```

```python
circuits = [(circuit_a_id,circuit_a), (circuit_b_id,circuit_b),]
circuits_decomposed = [(name, circuit.decompose(reps=10)) for name, circuit in circuits]
```

```python
for name, circuit in circuits_decomposed:
    display(circuit.draw("mpl", scale=0.5, fold=-1))
```

```python
results = evaluator.evaluate(circuits_decomposed)
```

```python
evaluator.print_metrics(results)
```
